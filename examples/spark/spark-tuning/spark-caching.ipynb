```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import *
```

```python
spark = (SparkSession
        .builder
        .appName("Spark caching")
        .config("spark.driver.memory", "8g")
        .getOrCreate())
```

```python
print("Configured:", spark.sparkContext.getConf().get("spark.driver.memory", "Not set (defaults to 1g)"))
```

```python
from pyspark.sql.functions import col

# Create a DataFrame with 10 million records
df = (spark.range(50_000_000).withColumn("square", col("id") * col("id")))
```

```python
df.cache()              # Mark the DataFrame for caching
```

```python
%time df.count()       # Materialize the cache
```

```python
%time df.count()       # read from cache
```

```python
spark.stop()
```
